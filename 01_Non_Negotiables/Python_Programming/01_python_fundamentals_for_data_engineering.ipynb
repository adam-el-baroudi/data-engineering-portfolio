```python
class Company():
    def __init__(self , cmpn_name , country):
        self.cmpn_name = cmpn_name
        self.country = country

    def infos(self):
        return(f"{self.cmpn_name} in {self.country}")

class managers():
    def __init__(self , m_name , m_age):
        self.m_name = m_name
        self.m_age  = m_age
    def infos(self):
        return (f"manager name {self.m_name} and he/she has {self.m_age} years old")

class Employee(managers,Company):
    def __init__(self , E_name , E_age , m_name , m_age , cmpn_name , country):
        managers.__init__(self,m_name,m_age)
        Company.__init__(self,cmpn_name,country)
        self.E_name = E_name
        self.E_age = E_age

    def infos(self):
        print(f"this emloyee {self.E_name} he/she has {self.E_age} years old and he/she works with manager : {managers.infos(self)} and the all work in {Company.infos(self)}")
        print("another methode")
        print(f"this emloyee {self.E_name} he/she has {self.E_age} years old and he/she works with manager : manager name {self.m_name} and he/she has {self.m_age} years old and the all work in {self.cmpn_name} in {self.country}")



E1 = Employee("adam" , 20 , "hassan" , "35" , "UM5" , "MA")
E1.infos()

print("===========multi level")
# Level 1: The "Grandparent" class (most general)
class Vehicle:
    def __init__(self, brand, year):
        self.brand = brand
        self.year = year

    def get_details(self):
        return f"{self.year} {self.brand}"


# Level 2: The "Parent" class (inherits from Vehicle)
class Car(Vehicle):
    def __init__(self, brand, year, num_doors):
        # Passes brand and year up to Vehicle
        super().__init__(brand, year) 
        self.num_doors = num_doors

    def get_details(self):
        # Grabs the string from Vehicle and adds to it
        return f"{super().get_details()} (Car with {self.num_doors} doors)"


# Level 3: The "Child" class (inherits from Car)
class ElectricCar(Car):
    def __init__(self, brand, year, num_doors, battery_capacity):
        # Passes brand, year, and num_doors up to Car
        super().__init__(brand, year, num_doors)
        self.battery_capacity = battery_capacity

    def get_details(self):
        # Grabs the string from Car (which already grabbed from Vehicle) and adds to it
        return f"{super().get_details()} | Battery: {self.battery_capacity}kWh"


# Let's test it out
my_ev = ElectricCar("Tesla", 2024, 4, 82)

print(my_ev.get_details())
```

Multi Threading 

```python
import time
import random
from concurrent.futures import ThreadPoolExecutor

tables = ['orders' , 'products' , 'customers' , 'reviews' , 'cancels']

def my_func(i):
    wait = random.randint(1,10)
    time.sleep(wait)
    print(f"I am {i} I took {wait} seconds")

for i in tables:
    my_func(i)
```

```python
#Multi Threading  is a runimg multiple chose in a same time 
#methode 1 with map
def my_func22(i):
    wait = 3
    time.sleep(wait)
    print(f"I am {i} I took {wait} seconds")
with ThreadPoolExecutor(max_workers=len(tables)) as executor:
    futures = executor.map(my_func22,tables)

print('===========')
#methode 2 with for
with ThreadPoolExecutor(max_workers=len(tables)) as executor:
    for i in tables:
        future = executor.submit(my_func22,i)
```
