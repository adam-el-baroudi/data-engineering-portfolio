# DATA WAREHOUSING

```sql
CREATE DATABASE IF NOT EXISTS orderDWH;
use orderDWH;
```

### Staging Layer

```sql
CREATE OR REPLACE TABLE orderDWH.stg_orders
AS
SELECT * FROM sales_new.orders
```

### Transformation

```sql
CREATE VIEW orderDWH.trans_sales
AS
SELECT * FROM orderDWH.stg_orders WHERE Quantity IS NOT NULL
```

```sql
SELECT * FROM orderdwh.trans_sales
```

#### DimCustomers

```sql
CREATE TABLE orderdwh.DimCustomers
(
    CustomerID INT,
    DimCustomerKey INT,
    CustomerName STRING,
    CustomerEmail STRING
)
```

```sql
CREATE OR REPLACE VIEW orderDWH.view_DimCustomers
AS
SELECT T.* , row_number() over(ORDER BY CustomerID) as DimCustomerKey FROM 
(
    SELECT 
        DISTINCT(CustomerID) as CustomerID,
        CustomerName,
        CustomerEmail

    FROM
        orderdwh.trans_sales
) AS T
```

```sql
SELECT * FROM orderdwh.view_DimCustomers
```

```sql
INSERT INTO orderdwh.dimcustomers (CustomerID, DimCustomerKey, CustomerName, CustomerEmail)
SELECT 
    CustomerID, 
    DimCustomerKey, 
    CustomerName, 
    CustomerEmail
FROM orderdwh.view_dimcustomers
```

#### DimProducts

```sql
CREATE TABLE orderdwh.DimProducts
(
    ProductID INT,
    ProductName STRING,
    ProductCategory STRING,
    DimProductKey INT
)
```

```sql
CREATE OR REPLACE VIEW orderDWH.view_DimProducts
AS
SELECT T.* , row_number() over(ORDER BY ProductID) as DimProductKey FROM 
(
    SELECT 
        DISTINCT(ProductID) as ProductID,
        ProductName,
        ProductCategory

    FROM
        orderdwh.trans_sales
) AS T
```

```sql
SELECT * FROM orderdwh.view_dimproducts
```

```sql
INSERT INTO orderdwh.dimproducts
Select * from orderdwh.view_dimproducts
```

```sql
SELECT * FROM orderdwh.dimproducts
```

#### DimRegions

```sql
CREATE TABLE IF NOT EXISTS orderdwh.DimRegions
(
    RegionID INT,
    RegionName STRING,
    Country STRING,
    DimRegionKey INT
)
```

```sql
CREATE OR REPLACE VIEW orderDWH.view_DimRegions
AS
SELECT T.* , row_number() over(ORDER BY RegionID) AS DimRegionKey FROM 
(
    SELECT DISTINCT
    RegionID,
    RegionName,
    Country
FROM
    orderdwh.trans_sales
) AS T
```

```sql
SELECT * FROM orderdwh.view_dimregions
```

```sql
INSERT INTO orderdwh.dimregions
SELECT * FROM orderdwh.view_dimregions
```

#### DimDate

```sql
CREATE OR REPLACE TABLE orderdwh.dimDate
(
    OrderDate DATE,
    DimDateKey INT
)
```

```sql
CREATE OR REPLACE VIEW orderdwh.view_dimDate
AS
SELECT T.* , row_number() over(ORDER BY OrderDate) AS DimDateKey FROM 
(
    SELECT DISTINCT
    OrderDate
FROM
    orderdwh.trans_sales
) AS T
```

```sql
SELECT * FROM orderdwh.view_DimDate
```

```sql
INSERT INTO orderdwh.dimdate
SELECT * FROM orderdwh.view_dimDate
```

#### Fact Table

```sql
CREATE OR REPLACE TABLE orderdwh.FactSales
(
    OrderID INT,
    Quantity DECIMAL,
    UnitPrice DECIMAL,
    TotalAmount DECIMAL,
    DimCustomerKey INT,
    DimProductKey INT,
    DimRegionKey INT,
    DimDateKey INT
)
```

```sql
SELECT 
    F.OrderID,
    F.Quantity,
    F.UnitPrice,
    F.TotalAmount,
    DC.DimCustomerKey,
    DP.DimProductKey,
    DD.DimDateKey,
    DR.DimRegionKey
FROM 
    orderdwh.trans_sales F
LEFT JOIN
    orderdwh.dimcustomers DC
ON
    F.CustomerID = DC.CustomerID
LEFT JOIN
    orderdwh.dimproducts DP
ON
    F.ProductID = DP.ProductID
LEFT JOIN
    orderdwh.dimdate DD
ON
    F.OrderDate = DD.OrderDate
LEFT JOIN
    orderdwh.dimregions DR
ON
    F.Country = DR.Country
```
